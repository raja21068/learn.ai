# Python Essentials: Functions & decisions
**Time:** 5 minutes  
**Objective:** Create reusable logic with conditions.

A function packages instructions under a name. Parameters accept inputs; return sends back a result. An if statement selects a branch depending on whether a condition is true.

> **Analogy:** A function resembles a small machine: supply inputs, and it returns a predictable output.


## Two ideas to keep in mind

- Define a function with `def name(parameter):`.
- Keep the function focused on one understandable task.


```python
def grade(score):
    if score >= 60:
        return 'Pass'
    return 'Try again'

print(grade(75))
```

## Five-minute hands-on exercise

Change the passing threshold to 70. Test the function with 69, 70, and 71.

### Quick knowledge check

What will `grade(40)` return?

- **A.** Pass
- **B.** Try again
- **C.** 40


<details>
<summary>Reveal answer: B</summary>

Because 40 is below 60, the if condition is false and execution reaches the second return.
</details>

**Takeaway:** Functions make programs easier to reuse, test, and explain.
