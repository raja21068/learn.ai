# Machine Learning: Train, validate, test
**Time:** 5 minutes  
**Objective:** Understand why a fair test uses unseen data.

A training set fits the model. A validation set helps select model settings. A test set estimates performance after choices are finalized. If you tune a model against the test set, its estimate becomes optimistic.

> **Analogy:** Practicing on last year’s exam solutions differs from sitting a new exam you have never seen.


## Two ideas to keep in mind

- Keep the test set out of the model-selection process.
- Prevent leakage: near-duplicate records or a person’s records should not cross splits when they create dependency.


```python
data = list(range(10))
train = data[:6]
validation = data[6:8]
test = data[8:]
print('Train:', train)
print('Validation:', validation)
print('Test:', test)
```
