# Agentic AI: Evaluation & guardrails
**Time:** 5 minutes  
**Objective:** Specify success, cost, and safety checks.

Agent evaluation should measure actual task outcomes, not only fluent explanations. Track success rate, incorrect tool calls, cost, latency, and unintended side effects. Use held-out cases and repeat runs where stochastic behavior matters. A successful benchmark result does not prove general reliability.

> **Analogy:** A delivery service should be graded on correct deliveries, delays, and damage—not on how confidently it describes routes.


## Two ideas to keep in mind

- Define tests with observable pass/fail criteria.
- Log tool actions and include failure cases, timeouts, and prompt-injection tests.


```python
runs = [True, False, True, True, False]
successes = sum(runs)
print('Success rate:', successes / len(runs))
print('Failures to inspect:', len(runs) - successes)
```

## Five-minute hands-on exercise

Change one failed run to a success, recompute the success rate, and state the change in percentage points.
