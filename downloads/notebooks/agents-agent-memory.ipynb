# Agentic AI: Short- & long-term memory
**Time:** 5 minutes  
**Objective:** Choose what state an agent needs to retain.

Working memory keeps relevant context for the current run. Persistent memory can store verified preferences, knowledge, or past results for future tasks. Memory can become stale, mistaken, sensitive, or inconsistent, so provenance and correction mechanisms matter.

> **Analogy:** A sticky note helps with today’s errands; a labeled archive helps you find trustworthy records later.


## Two ideas to keep in mind

- Keep temporary task state separate from persistent knowledge.
- Store only what is necessary and give users control over personal data.


```python
working_memory = {'task': 'draft checklist', 'step': 2}
knowledge = {'python': 'A general-purpose language'}
print('Current step:', working_memory['step'])
print('Known fact:', knowledge.get('python'))
```
