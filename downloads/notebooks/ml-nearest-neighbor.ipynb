# Machine Learning: Build a tiny classifier
**Time:** 5 minutes  
**Objective:** Use the nearest labeled example to make a prediction.

A one-nearest-neighbor classifier predicts the label of the closest training example under a chosen distance metric. It is a useful baseline, but feature scale and noise matter. Here we use one-dimensional distance for clarity.

> **Analogy:** If a new fruit has weight close to a known apple, a nearest-neighbor rule might assign the apple label.


## Two ideas to keep in mind

- Measure distance between a query and each training sample.
- Choose the label attached to the smallest distance.


```python
training = [(1, 'low'), (5, 'medium'), (9, 'high')]
query = 6
closest = min(training, key=lambda row: abs(row[0] - query))
print('Predicted label:', closest[1])
```

## Five-minute hands-on exercise
