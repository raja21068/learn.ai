# Python Essentials: Values & variables
**Time:** 5 minutes  
**Objective:** Print information and store a value.

Python runs instructions from top to bottom. A variable is a name that refers to a value. Strings represent text, while integers represent whole numbers. The print() function displays a result.

> **Analogy:** Think of a variable as a labeled drawer: the label is the name, and the value is what you put inside.


## Two ideas to keep in mind

- Assign a value using = (this is not a mathematical equality statement).
- Use f-strings to insert values into readable output.


```python
name = "Sam"
lessons = 3
print(f"Hi {name}! You finished {lessons} lessons.")
```

## Five-minute hands-on exercise

Change the name, increase the lesson count to 5, and predict the exact output before running it.

### Quick knowledge check

What does `score = 10` do?

- **A.** Compares score with 10
- **B.** Assigns 10 to score
- **C.** Prints 10


<details>
<summary>Reveal answer: B</summary>

The single = assigns a value. The comparison operator is ==.
</details>

**Takeaway:** A program becomes useful when you can name, store, and display information.
