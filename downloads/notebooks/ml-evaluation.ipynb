# Machine Learning: Metrics & overfitting
**Time:** 5 minutes  
**Objective:** Calculate accuracy and recognize its limits.

Accuracy is the fraction of correct predictions. It can be misleading when classes are imbalanced: predicting the majority class may look excellent while missing every rare event. Also compare training and held-out performance to detect overfitting.

> **Analogy:** A weather forecast saying “no rain” every day can appear accurate in a dry region, yet fail whenever rain matters.


## Two ideas to keep in mind

- Compute accuracy as correct predictions divided by total predictions.
- For imbalanced problems, examine precision, recall, and a confusion matrix too.


```python
true = [1, 0, 1, 0, 1]
pred = [1, 0, 0, 0, 1]
correct = sum(a == b for a, b in zip(true, pred))
print('Accuracy:', correct / len(true))
```
