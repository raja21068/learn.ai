# Agentic AI: The agent loop
**Time:** 5 minutes  
**Objective:** Identify observation, choice, action, and feedback.

An agent repeatedly observes state, decides on an action, executes it, and uses feedback to continue toward a goal. In LLM agents, the model may select tools while surrounding software controls execution and stopping criteria. A chatbot response alone is not necessarily an agent.

> **Analogy:** Think of a robot vacuum: it checks its surroundings, moves, checks again, and stops when its task is done.


## Two ideas to keep in mind

- The environment supplies observations after actions.
- Define a maximum number of steps and a clear success condition.


```python
items = ['plan', 'test', 'report']
completed = []
for step in range(3):
    observation = items[step]
    action = f'complete {observation}'
    completed.append(action)
    print('Observe:', observation, '-> Act:', action)
print('Done:', completed)
```
