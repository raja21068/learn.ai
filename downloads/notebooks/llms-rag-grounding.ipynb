# Large Language Models: Retrieval-augmented generation
**Time:** 5 minutes  
**Objective:** See how finding evidence differs from making up an answer.

Retrieval-augmented generation (RAG) retrieves passages from a data source and supplies them to a language model as context. Retrieval quality and citation grounding are critical. A retrieved document may be incorrect or malicious; RAG is not automatic fact checking.

> **Analogy:** Before answering a library question, look up relevant pages rather than guessing from memory.


## Two ideas to keep in mind

- Retrieve relevant evidence; do not assume the first match is sufficient.
- Distinguish quoted source content from generated conclusions.


```python
docs = {
    'python': 'Python is a programming language.',
    'rag': 'RAG supplies retrieved material as model context.',
}
query = 'What is RAG?'
term = 'rag' if 'rag' in query.lower() else 'python'
print('Source:', docs[term])
# Retrieval demo only: no language model is called.
```
