# Blockchain Basics: A shared ledger
**Time:** 5 minutes  
**Objective:** Explain what a blockchain records.

A blockchain is an append-oriented sequence of records grouped into blocks, where each block references a previous block through a cryptographic hash. Some blockchains use distributed consensus to agree on the current ledger. A blockchain does not make entered data automatically truthful.

> **Analogy:** Picture a notebook copied across many desks: everyone needs a rule for which new pages belong in the official copy.


## Two ideas to keep in mind

- Transactions record events or changes in state.
- Consensus and cryptographic checks address different trust problems.


```python
ledger = []
ledger.append({'from': 'A', 'to': 'B', 'amount': 3})
ledger.append({'from': 'B', 'to': 'C', 'amount': 1})
for event in ledger:
    print(event)
```
