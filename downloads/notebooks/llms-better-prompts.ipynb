# Large Language Models: Prompts & structured output
**Time:** 5 minutes  
**Objective:** Write a clear task with constraints and a checkable format.

A useful prompt states the objective, input, desired output format, and how to handle uncertainty. Examples can clarify expected behavior. Prompting improves reliability, but does not guarantee factuality or enforce security boundaries.

> **Analogy:** Giving instructions to a colleague works better when you specify both the task and what a good result looks like.


## Two ideas to keep in mind

- Make the output measurable: fields, length, source requirements.
- Treat text retrieved from untrusted sources as data, not instructions.


```python
task = 'Summarize this note in one sentence'
note = 'The group met Tuesday and agreed to test the model.'
prompt = f"""Task: {task}
Rules: Be factual; if unknown, say unknown.
Input: {note}
Output: one sentence
"""
print(prompt)
```
