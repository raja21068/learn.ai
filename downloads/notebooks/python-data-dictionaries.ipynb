# Python Essentials: Dictionaries & simple data
**Time:** 5 minutes  
**Objective:** Store structured information and compute a summary.

A dictionary stores values under meaningful keys, such as a person’s name or a score. Dictionaries are useful for records, configuration, and data analysis. You can combine dictionaries with basic arithmetic.

> **Analogy:** A dictionary is a filing cabinet: each labeled folder leads to its corresponding value.


## Two ideas to keep in mind

- Read a value with `record["key"]`.
- Use `sum()` and `len()` to compute a simple average.


```python
student = {'name': 'Ava', 'scores': [80, 90, 100]}
average = sum(student['scores']) / len(student['scores'])
print(student['name'], 'average:', average)
```

### Quick knowledge check

Which structure is best for `{"name": "Ava", "age": 19}`?

- **A.** A list
- **B.** A dictionary
- **C.** A float


<details>
<summary>Reveal answer: B</summary>

</details>

**Takeaway:** When your data has named fields, dictionaries are often clearer than parallel lists.
