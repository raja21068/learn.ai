# Agentic AI: Planning & tools
**Time:** 5 minutes  
**Objective:** Break a task into bounded, observable actions.

Planning decomposes a goal into steps and revises them when evidence changes. Tools connect agents to code, files, browsers, and databases. A robust plan checks preconditions, handles failure, and avoids performing irreversible actions without authorization.

> **Analogy:** Planning a trip means choosing stops, checking transport, and adapting when a train is cancelled.


## Two ideas to keep in mind

- Choose small steps with verifiable outputs.
- Keep tool permissions narrow and distinguish read-only actions from writes.


```python
plan = [
    {'task': 'read file', 'requires_approval': False},
    {'task': 'delete file', 'requires_approval': True},
]
for step in plan:
    state = 'ask user' if step['requires_approval'] else 'can inspect'
    print(step['task'], '=>', state)
```
