# Blockchain Basics: Hashes & tamper evidence
**Time:** 5 minutes  
**Objective:** Observe how a hash changes when input changes.

A cryptographic hash transforms arbitrary data into a fixed-size digest. SHA-256 is deterministic, and tiny input changes usually produce very different digests. A hash is not encryption: there is no decryption key for recovering the original message.

> **Analogy:** A hash resembles a fingerprint of data, except it is computed mathematically and not uniquely guaranteed for all possible inputs.


## Two ideas to keep in mind

- Use a canonical encoding before hashing structured records.
- Hashes help expose edits, but alone do not prove authorship or establish consensus.


```python
import hashlib
for text in ['hello', 'Hello']:
    digest = hashlib.sha256(text.encode()).hexdigest()
    print(text, digest[:16])
```

## Five-minute hands-on exercise

Hash your own two messages that differ by one character; compare their first 16 digest characters.
