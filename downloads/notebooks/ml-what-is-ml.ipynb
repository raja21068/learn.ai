# Machine Learning: What is machine learning?
**Time:** 5 minutes  
**Objective:** Tell conventional code apart from learning from examples.

Traditional programs follow rules written by people. In supervised machine learning, an algorithm fits a model to input-output examples so it can predict outcomes for new inputs. Learning a pattern is not the same as understanding causes.

> **Analogy:** Instead of writing every rule for recognizing fruit, show the model examples with measurable features and labels.


## Two ideas to keep in mind

- Features are the input measurements; labels are target answers.
- A model learns from examples, then predicts for unseen examples.


```python
examples = [(1, 2), (2, 4), (3, 6)]
# Illustration: all examples follow y = 2 * x.
def predict(x):
    return 2 * x
print('Prediction for 4:', predict(4))
```
