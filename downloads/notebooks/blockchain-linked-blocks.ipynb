# Blockchain Basics: Make a toy blockchain
**Time:** 5 minutes  
**Objective:** Connect two blocks using a previous hash.

Each block can store a payload and a digest of the prior block. When a previous block changes, its hash no longer matches the next block’s stored link. This demonstration omits consensus, signatures, networking, and persistence; it is not a secure cryptocurrency.

> **Analogy:** It is like numbering pages and including the previous page’s checksum on the next page.


## Two ideas to keep in mind

- Serialize records consistently before hashing.
- Linking blocks detects inconsistency, but an attacker able to rewrite every block may recompute all hashes.


```python
import hashlib, json

def digest(record):
    raw = json.dumps(record, sort_keys=True).encode()
    return hashlib.sha256(raw).hexdigest()

first = {'prev': 'GENESIS', 'data': 'A pays B'}
second = {'prev': digest(first), 'data': 'B pays C'}
print(second['prev'] == digest(first))
```

## Five-minute hands-on exercise

Change first["data"] after creating second. Re-run the link check and explain why it now fails.

### Quick knowledge check

If an attacker rewrites the first block and all following hashes, what is still needed to resist the attack?

- **A.** A larger font
- **B.** An external trust or consensus mechanism
- **C.** More print statements


<details>
<summary>Reveal answer: B</summary>

Hash links alone are insufficient against someone who can recompute the entire chain.
</details>

**Takeaway:** Security claims require an explicit threat model, not just a hash function.
