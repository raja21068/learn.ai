# Large Language Models: Tokens & context
**Time:** 5 minutes  
**Objective:** Understand the basic units used by language models.

Language models operate on tokens, which may represent words, word pieces, punctuation, or other units. They predict sequences conditional on context. A model’s context window limits what it can directly attend to in one request, and tokenization is model-specific.

> **Analogy:** Think of tokens as the tiles used to construct sentences; tile boundaries do not always match spaces.


## Two ideas to keep in mind

- Tokens are not the same as words.
- A convincing sentence is not evidence that a statement is true.


```python
text = 'Models predict token sequences.'
# This is a word split for teaching, NOT real model tokenization.
words = text.split()
print(words)
print('Word count:', len(words))
```
