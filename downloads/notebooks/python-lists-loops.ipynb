# Python Essentials: Lists & loops
**Time:** 5 minutes  
**Objective:** Repeat work over a collection.

A list stores several items in order. A for loop visits each item once. This avoids writing the same instruction again and again. Use len() to count items.

> **Analogy:** A list is a shopping basket. A for loop picks up each item in the basket and checks it.


## Two ideas to keep in mind

- Lists use square brackets; positions begin at 0.
- Use `for item in items:` with an indented body.


```python
scores = [70, 85, 90]
for score in scores:
    print('Score:', score)
print('Count:', len(scores))
```

## Five-minute hands-on exercise

Add 100 to the scores list and print the new item count and total using sum().
