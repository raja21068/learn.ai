# Blockchain Basics: Consensus & smart contracts
**Time:** 5 minutes  
**Objective:** Distinguish shared agreement from programmed rules.

Consensus rules help network participants agree on an ordered history despite failures or adversarial behavior under stated assumptions. A smart contract is code whose state transitions are executed according to platform rules. Contracts can still contain bugs and depend on unreliable external inputs.

> **Analogy:** Consensus decides which pages belong in the shared notebook; a smart contract is a rule written on those pages.


## Two ideas to keep in mind

- Different consensus approaches make different tradeoffs in speed, cost, and trust assumptions.
- Audits, tests, and careful oracle design matter for real contracts.


```python
balance = {'alice': 10, 'bob': 0}
def transfer(amount):
    if amount <= 0 or balance['alice'] < amount:
        return False
    balance['alice'] -= amount
    balance['bob'] += amount
    return True
print(transfer(4), balance)
```

## Five-minute hands-on exercise

Attempt transfers of 15 and then 0. Check that both fail and balances remain unchanged.
