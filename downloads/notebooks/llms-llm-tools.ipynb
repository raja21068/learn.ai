# Large Language Models: Tool calling
**Time:** 5 minutes  
**Objective:** Separate model decisions from actual tool execution.

Tool calling lets a model request a structured action such as searching, calculating, or accessing a database. The application validates arguments and runs the tool. Permission checks and human approval are needed for sensitive side effects.

> **Analogy:** A model can ask for a calculator, but the calculator—not the model—performs the arithmetic.


## Two ideas to keep in mind

- Define a tool with a narrow input schema.
- Validate and log execution; never blindly run model-produced shell commands.


```python
def calculator(a, b, operation):
    if operation not in {'add', 'multiply'}:
        raise ValueError('Unsupported operation')
    return a + b if operation == 'add' else a * b

request = {'a': 8, 'b': 7, 'operation': 'multiply'}
print(calculator(**request))
```

## Five-minute hands-on exercise

Call the allowed add operation, then try an unsupported operation and observe the validation error.
